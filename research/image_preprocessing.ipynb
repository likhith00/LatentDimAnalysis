```python
from torch.utils.data import DataLoader, Subset, ConcatDataset
from src.load.image import (
    create_image_transform,
    load_dataset_tvt,
    get_dataset_mapping
)
from src.data.image import preprocess_torchvision_dataset
```

```python
IMAGE_RESIZE_CONFIG = {
    "flowers102": (52, 52),
    "food101": (56, 56),
    "cifar100": (32, 32),
    "gtsrb": (32, 32),
}
```

```python
dataset_name = "flowers102"
dataset_mapping = get_dataset_mapping(dataset_name=dataset_name)

transform = create_image_transform(
    IMAGE_RESIZE_CONFIG[dataset_name]
)

train_dataset, val_dataset, test_dataset = load_dataset_tvt(
    name=dataset_name,
    dataset_mapping=dataset_mapping,
    root="../root",
    transform=transform
)

full_dataset = ConcatDataset([
    train_dataset,
    val_dataset,
    test_dataset
])
```

```python
prep_data = preprocess_torchvision_dataset(
    dataset=full_dataset,
    max_samples=10000,
    batch_size=128,
    num_workers=4
)
```
